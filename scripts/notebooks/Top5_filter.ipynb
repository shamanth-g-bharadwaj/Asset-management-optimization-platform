```python
# Step 1: Import necessary libraries
import pandas as pd

# Step 2: Load your dataset
file_path = r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\Datasets\US Stocks\NASDAQ_Combined_Dataset.csv"
df = pd.read_csv(file_path)

# Step 3: Convert 'Date' column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Step 4: Define your top 5 stocks
top_5_stocks = ['ceg', 'NVDA', 'PLTR', 'arm', 'amd']  # Replace with your actual top 5

# Step 5: Build the list of required columns
selected_columns = ['Date'] + [f"{ticker}_Close/Last" for ticker in top_5_stocks]

# Step 6: Filter the dataset to only include the selected columns
filtered_df = df[selected_columns].copy()

# Step 7: Convert Close/Last columns to float (removing $ if necessary)
for ticker in top_5_stocks:
    col = f"{ticker}_Close/Last"
    filtered_df[col] = (
        filtered_df[col]
        .astype(str)
        .str.replace('$', '', regex=False)
        .astype(float)
    )

# Step 8: Display the first few rows to verify
filtered_df.head()
```
