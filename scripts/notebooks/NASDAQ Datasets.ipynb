```python
import pandas as pd
import os

# Set the folder path where your CSV files are stored
folder_path = r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\NASDAQ Datasets"

# Get all CSV files
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
print(f"Found {len(csv_files)} CSV files.")

# Initialize an empty dataframe for merging
merged_df = None

for file in csv_files:
    stock_name = os.path.splitext(file)[0]  # Get the filename without extension
    file_path = os.path.join(folder_path, file)

    # Read CSV
    df = pd.read_csv(file_path)

    # Ensure Date column is datetime
    df['Date'] = pd.to_datetime(df['Date'])

    # Rename other columns to include the stock name
    df_renamed = df.rename(columns={
        'Close/Last': f'{stock_name}_Close/Last',
        'Volume': f'{stock_name}_Volume',
        'Open': f'{stock_name}_Open',
        'High': f'{stock_name}_High',
        'Low': f'{stock_name}_Low'
    })

    # If first file, initialize merged_df
    if merged_df is None:
        merged_df = df_renamed
    else:
        # Merge on 'Date' column
        merged_df = pd.merge(merged_df, df_renamed, on='Date', how='outer')

# Sort by date
merged_df.sort_values(by='Date', inplace=True)

# Save the merged dataset
output_path = os.path.join(folder_path, 'merged_stock_data.csv')
merged_df.to_csv(output_path, index=False)
print(f"Merged dataset saved at: {output_path}")
```
