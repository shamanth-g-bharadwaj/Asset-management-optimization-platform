# Analyzing and Forecasting NASDAQ Stocks

**Importing Libraries and Dataset**

```python
import pandas as pd

# Load the NASDAQ dataset
file_path = r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\Datasets\US Stocks\NASDAQ Dataset.csv"

# Try loading and displaying basic info
df_stocks = pd.read_csv(file_path)

# Quick overview
print(df_stocks.shape)
print(df_stocks.columns)
df_stocks.head()
```

```python
# Ensure 'Date' is parsed and set as the datetime index
df_stocks['Date'] = pd.to_datetime(df_stocks['Date'], errors='coerce')  # parse and handle any bad dates
df_stocks.dropna(subset=['Date'], inplace=True)                         # drop rows with invalid dates
df_stocks.set_index('Date', inplace=True)
df_stocks.sort_index(inplace=True)  # optional: keep dates in order
```

```python
# Remove '$' and commas, then convert to numeric
for col in df_stocks.columns:
    if any(col.endswith(suffix) for suffix in ['_Open', '_High', '_Low', '_Close/Last']):
        df_stocks[col] = (
            df_stocks[col]
            .astype(str)
            .str.replace('$', '', regex=False)
            .str.replace(',', '', regex=False)
        )
        df_stocks[col] = pd.to_numeric(df_stocks[col], errors='coerce')
        
```

```python
# Print all column names
for i, col in enumerate(df_stocks.columns):
    print(f"{i}: {col}")
```

```python
# Drop all columns containing '_Volume' in their name
df_stocks = df_stocks.loc[:, ~df_stocks.columns.str.contains('_Volume')]
```

```python
# Print all column names
for i, col in enumerate(df_stocks.columns):
    print(f"{i}: {col}")
```

```python
# Rename all columns ending with '_Close/Last' to '_Close'
df_stocks.rename(columns=lambda col: col.replace('_Close/Last', '_Close') if col.endswith('_Close/Last') else col, inplace=True)
```

```python
# Print all column names
for i, col in enumerate(df_stocks.columns):
    print(f"{i}: {col}")
```

```python
# Extracting all stock prefixes based on columns ending with '_Close'
close_columns = [col for col in df_stocks.columns if col.endswith('_Close')]
stock_prefixes = [col.replace('_Close', '') for col in close_columns]

# Confirm sample
print(f"Found {len(stock_prefixes)} stocks.")
print(stock_prefixes[:5])
```

```python
from statsmodels.tsa.api import VAR
import pandas as pd
import numpy as np

# Container to store results
forecast_all_df = []
forecast_days = 30
forecast_dates = pd.date_range(start=df_stocks.index[-1] + pd.Timedelta(days=1), periods=forecast_days, freq='B')

for stock in stock_prefixes:
    try:
        # Use only _Open and _Close
        cols = [f"{stock}_Open", f"{stock}_Close"]
        stock_df = df_stocks[cols].dropna()

        if stock_df.shape[0] < 40:
            continue

        # Difference the data
        stock_diff = stock_df.diff().dropna()

        # Fit VAR model
        model = VAR(stock_diff)
        results = model.fit(maxlags=15, ic='aic')

        # Forecast differenced values
        lag_order = results.k_ar
        forecast_input = stock_diff.values[-lag_order:]
        forecast_diff = results.forecast(y=forecast_input, steps=forecast_days)

        # Create DataFrame for forecasted diffs
        forecast_diff_df = pd.DataFrame(forecast_diff, columns=cols)

        # Reconstruct actual values
        last_known = stock_df.iloc[-1]
        forecast_actual = forecast_diff_df.cumsum() + last_known

        # Store only the Close forecast
        forecast_stock = pd.DataFrame({
            'Date': forecast_dates,
            'Forecasted_Close': forecast_actual[f"{stock}_Close"],
            'Stock': stock
        })

        forecast_all_df.append(forecast_stock)

    except Exception as e:
        print(f"❌ Skipped {stock}: {e}")
        continue

# Combine all into one DataFrame
forecast_all_df = pd.concat(forecast_all_df).reset_index(drop=True)
```

```python
# STEP: Pivot forecast data so each stock becomes a column
pivot_forecast_df = forecast_all_df.pivot(index='Date', columns='Stock', values='Forecasted_Close')

# Reset index so Date is a column
pivot_forecast_df.reset_index(inplace=True)

# Export to CSV for Tableau
pivot_forecast_df.to_csv(r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\Datasets\US Stocks\NASDAQ Dataset_Draft1.csv", index=False)

# Optional: Preview first few rows
pivot_forecast_df.head()
```
