# Analysis of UK Inflation: Trends in Food, Energy, and Housing Indices

**Importing Libraries & Datasets**
(Filtering Data to Procure Relevant Time Frame)

```python
import pandas as pd

# Load Food at Home CPI
food_df = pd.read_csv(r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\Final Files\Final Datasets\UK Inflation Raw\UK_Food_Index.csv", skiprows=501, header=None)
# Load Energy CPI
energy_df = pd.read_csv(r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\Final Files\Final Datasets\UK Inflation Raw\UK_Energy_Index.csv", skiprows=518, header=None)
# Load Housing CPI
housing_df = pd.read_csv(r"D:\Masters Study Abroad\BA 2nd Sem\IS6611 Applied Research in Business Analytics\IS6611 IT Artefact V3\Final Files\Final Datasets\UK Inflation Raw\UK_Housing_Index.csv", skiprows=501, header=None)
```

```python
# Display to confirm
display(food_df.head())
display(energy_df.head())
display(housing_df.head())
```

**Assigning Columns Names**

```python
# Assign column names
food_df.columns = ['Date', 'Food_Index']
energy_df.columns = ['Date', 'Energy_Index']
housing_df.columns = ['Date', 'Housing_Index']

# Display to confirm
display(food_df.head())
display(energy_df.head())
display(housing_df.head())
```

**Converting the 'Date' column in all three DataFrames**

```python
food_df['Date'] = pd.to_datetime(food_df['Date'], format='%Y %b')
energy_df['Date'] = pd.to_datetime(energy_df['Date'], format='%Y %b')
housing_df['Date'] = pd.to_datetime(housing_df['Date'], format='%Y %b')

# Display to confirm correct datetime conversion
display(food_df.head())
display(energy_df.head())
display(housing_df.head())
```

**Dataset Consolidation**

```python
# Merge food and energy first
UK_inflation_df = pd.merge(food_df, energy_df, on='Date', how='inner')

# Then merge the result with housing
UK_inflation_df = pd.merge(UK_inflation_df, housing_df, on='Date', how='inner')

# Display the merged DataFrame
display(UK_inflation_df.head())
```

**Computing Percentage Growth in Inflation Indices**
(Calculating percentage growth for the Energy index only, since the other indices are already expressed as percentage growth.)

```python
# Get the base (Jan 2015) energy index value
base_energy = UK_inflation_df['Energy_Index'].iloc[0]

# Convert Energy Index to % growth from Jan 2015
UK_inflation_df['Energy_Growth_%'] = ((UK_inflation_df['Energy_Index'] - base_energy) / base_energy * 100).round(2)

# Rename the existing columns Food_Index and Housing_Index
UK_inflation_df.rename(columns={
    'Food_Index': 'Food_Growth_%',
    'Housing_Index': 'Housing_Growth_%'
}, inplace=True)
```
